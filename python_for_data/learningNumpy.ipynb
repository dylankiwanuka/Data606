## Iterating through arrays

```python
# if 1d array
arr1 = np.array([10,20,30])
for x in arr1:
    print(x)
```

```python
# if its an n-dimensional array we need to first iterate through the dimensions first
for dim in arr3:
    for num in dim:
        print(num)
```

```python
# nditer = nd iterate  allows you to iterate without having to use nested loops : gives all elements 
for n in np.nditer(arr3):
    print(n)
```
