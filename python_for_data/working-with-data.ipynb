### CSV Files in Pandas

Download the iris dataset through the terminal. Make sure it is in the same folder as this notebook

```python
# using read_csv
iris = pd.read_csv("iris.csv")
iris
```

```python
# Head
iris.head()
```

```python
# Writing DF to a csv file
iris.to_csv("iris_file_rewrite")
```

```python
#group by splits your table into groups and then does a calculation on that group
#group the rows by species, then take the average of each column
iris.groupby("species").mean()
```

```python
#Iterating through the DataFrame

# iterrows() hands us the table ONE ROW at a time, like flipping through flower cards
for i, row in iris.iterrows():
    # each loop gives us two things:
    #   i   = the row's label (0, then 1, then 2...)
    #   row = that whole row (a mini Series with all 5 columns)

    # pick one value from the row by its column name, then print it with the row number
    print(i, row["species"])   # e.g. 0 setosa
```

```python
# Selecting multiple values when iterating:
for i, row in iris.iterrows():
    # double brackets = a list of column names, so we get several values back
    print(i, row[["species", "petal_length", "petal_width"]])
```

```python
# Using Describe()
iris.describe()
```

```python
#Ranking petal length in asc order:

iris["Rank Petal Length"] = iris["petal_length"].rank(ascending=0)
iris
```

```python
iris.groupby("species").mean()
```

```python
#Creating a new DataFrame for each species

# Make a new table for each species by keeping only the matching rows
# iris["species"] == "setosa" gives True/False for every row, iris[...] keeps the True ones
setosa = iris[iris["species"] == "setosa"]
versicolor = iris[iris["species"] == "versicolor"]
virginica = iris[iris["species"] == "virginica"]

# Show the first 5 rows of the virginica table
virginica.head()
```

```python
#Creating sepal and petal info from iris
sepal_info = iris[["sepal_length", "sepal_width", "species"]]
petal_info = iris[["petal_length", "petal_width", "species"]]
sepal_info.head()
```

```python
#Loading a Json into a dataframe
iris_json = pd.read_json("iris.json")
iris_json.head()
```

```python
# Merging sepal_info and petal_info
# Join the two tables side by side, matching rows that have the same species
mergedDF = pd.merge(sepal_info, petal_info, on = "species")
mergedDF.head()
```

```python
# concat glues tables together. It doesn’t match anything, it just sticks them in a line.
# essentially stacks dataframes
# Stack petal_info underneath sepal_info (adds rows, one table on top of the other)
stacked = pd.concat([sepal_info, petal_info])
stacked
```
