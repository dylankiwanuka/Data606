### 2. Weather Data

```python
#Days above 25
temperatures = np.array([18, 21, 25, 17, 16, 23, 28, 30, 22, 19, 15, 24, 27, 20])
for item in temperatures:
    if item > 25:
        print(item)
```

```python
#Days below 20
for item in temperatures:
    if item < 20:
        print(item)
```

```python
#Avg Temp
temperatures.mean()
```

```python
#Hottest Day
temperatures.max()
```

```python
#Coldest Day
temperatures.min()
```

```python
#Days Exceeded avg temp
for item in temperatures:
    if item > temperatures.mean():
        print (item)
```

## 4. Random Customer purchases

```python
# Make a random number generator
rng = np.random.default_rng()

# 15 random purchases between £1 and £500
purchases = rng.uniform(low=1, high=500, size=12)
purchases
```

```python
# Reshape essentially puts an array into a given dimension size: total items must stay the same
purchases.reshape(3,4)
```

```python
# Transpose flips an array so its rows become columns and its columns become rows
grid = purchases.reshape(3, 4)
grid.T
```
